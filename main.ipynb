```python
import subprocess

def unload_all_models():
    """ 
    Gestiona la descarga de cualquier modelo cargado actualmente 
    usando la CLI 'lms' de LM Studio para liberar la VRAM. 
    """
    print("ðŸ§¹ Iniciando la descarga de modelos activos...")
    try:
        # Se ejecuta el comando lms unload --all para purgar la memoria
        cmd = ["lms", "unload", "--all"]
        result = subprocess.run(cmd, capture_output=True, text=True, check=False)
        
        # Se verifica el cÃ³digo de retorno para confirmar el Ã©xito de la operaciÃ³n
        if result.returncode == 0:
            print("  âœ… Todos los modelos han sido descargados exitosamente.")
        else:
            print(f"  âš ï¸ Error descargando modelos via CLI. Detalles: {result.stderr.strip()}")
            
    except FileNotFoundError:
        print("  âš ï¸ Error: No se encontrÃ³ el comando 'lms'. AsegÃºrese de que la CLI de LM Studio estÃ© instalada y configurada en las variables de entorno (PATH).")
    except Exception as e:
        print(f"  âš ï¸ Error inesperado en la gestiÃ³n de descarga: {e}")

# Ejemplo de uso:
unload_all_models()
```

```python
import os
import glob
import subprocess
import shutil

# === 1. ConfiguraciÃ³n de Carpetas ===
carpeta_audios = r"Resultados\Audios"
carpeta_json = r"Resultados\Diario"
carpeta_auxiliares = r"Auxiliares"

# Creamos las carpetas de destino en caso de que aÃºn no existan
os.makedirs(carpeta_json, exist_ok=True)
os.makedirs(carpeta_auxiliares, exist_ok=True)

# Token y comando de base
hf_token = "TU_HF_TOKEN_AQUI"
comando_base = r"C:\Users\Usuario\anaconda3\condabin\conda.bat activate whisperx_5090 && cd common && python -m whisperx"

# Convertimos las carpetas de destino a rutas absolutas
output_dir_abs = os.path.abspath(carpeta_auxiliares)
json_dir_abs = os.path.abspath(carpeta_json)

# === 2. BÃºsqueda de Audios ===
# Puedes agregar iteraciones para extensiones .mp3 si tambiÃ©n manejas esos formatos cambiando "*.wav"
archivos_audio = glob.glob(os.path.join(carpeta_audios, "*.wav"))

if not archivos_audio:
    print(f"No se encontraron archivos .wav en la carpeta: {carpeta_audios}")

# === 3. Bucle de Procesamiento ===
for audio_relativo in archivos_audio:
    audio_path = os.path.abspath(audio_relativo)
    
    print(f"\n" + "="*50)
    print(f"â–¶ï¸ Procesando: {os.path.basename(audio_path)}")
    print("="*50)
    
    # Construir el comando. Usamos --output_dir para indicar dÃ³nde dejar TODO inicialmente
    cmd = (
        f'{comando_base} "{audio_path}" '
        f'--language es --model large-v3 --batch_size 64 --threads 16 '
        f'--compute_type float16 --hf_token {hf_token} --diarize '
        f'--diarize_model pyannote/speaker-diarization-3.1 '
        f'--output_dir "{output_dir_abs}"'
    )
    
    # Ejecutamos en terminal localmente (puedes ver la barra de progreso en el cell output)
    try:
        # Pasa shell=True para evaluar logica de "&&" correctamente
        subprocess.run(cmd, shell=True, check=True)
        print(f"\nâœ… Terminado exitosamente: {os.path.basename(audio_path)}")
        
        # === 4. Movimiento de archivos ===
        # Obtener el nombre original base (Ej: "1_2_chau")
        nombre_base = os.path.splitext(os.path.basename(audio_path))[0]
        
        # Ruta donde WhisperX dejÃ³ el json y hacia donde irÃ¡
        json_generado = os.path.join(output_dir_abs, f"{nombre_base}.json")
        json_destino = os.path.join(json_dir_abs, f"{nombre_base}.json")
        
        if os.path.exists(json_generado):
            # Movemos el JSON a Resultados/diario, el .srt, .txt, .tsv, etc se quedan en Auxiliares
            shutil.move(json_generado, json_destino)
            print(f"ðŸšš Â¡JSON movido exitosamente a: {json_destino}!")
        else:
            print(f"âš ï¸ Advertencia: No se encontrÃ³ el JSON generado para este fragmento: {json_generado}")

    except subprocess.CalledProcessError as e:
        print(f"\nâŒ Se produjo un error procesando {os.path.basename(audio_path)}. Revisa los logs de arriba.")
        
print("\nðŸ Â¡Procesamiento por la carpeta de lotes finalizado!")
```

```python
import os
import re

def obtener_ids_de_carpeta(ruta_carpeta):
    """
    Escanea la carpeta y extrae el 'id' de archivos con formato 'id_x.ext'.
    Retorna un conjunto (set) de identificadores.
    """
    ids = set()
    # ExpresiÃ³n regular para capturar lo que estÃ¡ antes del primer guion bajo
    # Ejemplo: en '123_documento.txt', captura '123'
    patron = re.compile(r'^([^_]+)_')

    if not os.path.exists(ruta_carpeta):
        print(f"Error: La ruta {ruta_carpeta} no existe.")
        return ids

    for nombre_archivo in os.listdir(ruta_carpeta):
        coincidencia = patron.match(nombre_archivo)
        if coincidencia:
            ids.add(coincidencia.group(1))
            
    return ids

def comparar_carpetas(ruta_a, ruta_b):
    ids_a = obtener_ids_de_carpeta(ruta_a)
    ids_b = obtener_ids_de_carpeta(ruta_b)

    # Diferencia de conjuntos
    faltan_en_b = ids_a - ids_b
    faltan_en_a = ids_b - ids_a

    print(f"--- Resultados de la ComparaciÃ³n ---")
    print(f"IDs en Carpeta A: {len(ids_a)}")
    print(f"IDs en Carpeta B: {len(ids_b)}")
    print(f"\nIDs que estÃ¡n en A pero faltan en B: {list(faltan_en_b)}")
    print(f"IDs que estÃ¡n en B pero faltan en A: {list(faltan_en_a)}")

# Ejemplo de uso
if __name__ == "__main__":
    dir_1 = r"Resultados\Fragmentos"
    dir_2 = r"Resultados/Atomico"
    comparar_carpetas(dir_1, dir_2)
```
